```python
import pandas as pd
from openai import OpenAI
import time
```

```python
data = pd.read_csv("processed_data.csv")
data.head()
```

```python
data.columns
```

```python
client = OpenAI(
    base_url = "https://openrouter.ai/api/v1",
    api_key = "YOUR_API_KEY"
)
```

```python
def arabic_to_english_translation(text):

    prompt = f"""
Translate the following Arabic text into English.

Rules:
- Only return the English translation.
- Do not add extra information.
- Keep the meaning accurate.

Arabic text:
{text}
"""
    try:
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            messages=[
                {
                    "role": "system",
                    "content": "You are a professional Arabic to English translator."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=150,
            temperature=0.3
        )

        return response.choices[0].message.content.strip()

    except Exception as e:
        print("Error:", e)
        return ""
```

```python
sample_question = data["Question_for_translation"].iloc[10]
sample_answer = data["Answer_for_translation"].iloc[10]

print("Arabic Question:")
print(sample_question)

print("\nEnglish Question:")
print(arabic_to_english_translation(sample_question))

print("\nArabic Answer:")
print(sample_answer)

print("\nEnglish Answer:")
print(arabic_to_english_translation(sample_answer))
```

```python
examples = data.head(5).copy()

examples["Question_EN"] = examples["Question_for_translation"].apply(
    arabic_to_english_translation
)

examples["Answer_EN"] = examples["Answer_for_translation"].apply(
    arabic_to_english_translation
)

examples[
    [
        "Question_for_translation",
        "Question_EN",
        "Answer_for_translation",
        "Answer_EN"
    ]
]
```
